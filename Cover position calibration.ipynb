# Cover position calibration

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

## Measurements

```python
df = pd.DataFrame(data={
    'somfy': [100, 75, 50, 25],
    'truth': [100, 85, 64, 42]
})

df
```

```python
df.plot(x="somfy", y="truth")

plt.grid()
plt.show()
```

## Interpolation

```python
result = linregress(df["somfy"], df["truth"])
y_pred = result.slope * df["somfy"] + result.intercept
```

```python
plt.scatter(df["somfy"], df["truth"])
plt.plot(df["somfy"], y_pred)

plt.grid()
plt.show()
```

## Compute new positions

```python
for p in range(20, 100, 10):
    print(f'{p}: {result.slope * p + result.intercept}')
```
